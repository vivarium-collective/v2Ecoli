```python
# %%
# --- Query 1: bulk molecules only ---

import duckdb
import numpy as np
import pandas as pd
from duckdb import DuckDBPyConnection


query_dict = {
    "experiment_id": "default",
    "variant": 0,
    "lineage_seed": 0,
    "generation": 1,
}

query = f"""
    SELECT bulk__id, global_time FROM read_parquet(".pbg/parquet-runs/{query_dict["experiment_id"]}/history/*/*/*/*/*/*.pq", hive_partitioning=true)
    WHERE variant={query_dict["variant"]}
    AND lineage_seed={query_dict["lineage_seed"]}
    AND generation={query_dict["generation"]}
    ORDER BY global_time
"""
db_bulk = duckdb.sql(query).df()
# Stack bulk column into 2D numpy array: rows=timesteps, cols=molecules
bulk_state_mtx = np.stack(db_bulk["bulk__id"].values)

print(bulk_state_mtx)

```
